# Evaluate Pain Classifier

Loads a saved logistic regression classifier and reports precision, recall, and F1 on a small labeled evaluation set.

```python
import pandas as pd

eval_examples = [
    ("I manually reconcile refunds in a spreadsheet every month", 1),
    ("Can someone recommend a GitHub Actions log analyzer?", 1),
    ("Dashboards are weird", 0),
    ("I like building side projects", 0),
]
eval_df = pd.DataFrame(eval_examples, columns=["text", "label"])
eval_df
```

```python
import joblib
from sentence_transformers import SentenceTransformer
from sklearn.metrics import classification_report, confusion_matrix

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
clf = joblib.load("pain_classifier.joblib")
X = model.encode(eval_df.text.tolist(), normalize_embeddings=True)
pred = clf.predict(X)
print(classification_report(eval_df.label, pred))
print(confusion_matrix(eval_df.label, pred))
```
